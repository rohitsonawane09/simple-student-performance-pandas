### Create a new column score_level
"High" if total_score ≥ 90 , "Medium" if 70 ≤ total_score < 90, "Low" if total_score < 70

```python
def level(x):
  if x >= 90:
    return "High"
  elif x >= 70 and x < 90:
    return "Medium"
  elif x < 70:
    return "Low"

df["Scoring Level"] = df['total_score'].apply(level)
df

```

### Which student has the highest total_score?

```python
high = df[df['total_score'] == df['total_score'].max()].count()['student_id']
total = df.shape[0]
perc = (high*100)/total
perc
```

### Create a new column study_category
"Light" if weekly_self_study_hours < 10,
"Moderate" if 10 ≤ hours ≤ 20,
"Heavy" if hours > 20

```python
def category(x):
  if x < 10 :
    return "Light"
  elif x <= 20:
    return "Moderate"
  else:
    return "Heavy"

df['study category'] = df['weekly_self_study_hours'].apply(category)
```
